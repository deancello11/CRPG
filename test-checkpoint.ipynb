```python
# GEN AI - TEXT RPG POWERED BY A TRANSFORMER
#
# This game uses GPT-2 to generate descriptive text
# for each scene, while your code keeps track of the
# actual game state (health, choices, etc).

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# ==================================================
# LOAD TOKENIZER AND MODEL
# ==================================================
tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")

# ==================================================
# TEXT GENERATION HELPER
# ==================================================
def generate_text(prompt, max_new_tokens=40):
    """Feed a prompt to GPT-2 and get back a short continuation."""
    inputs = tokenizer(prompt, return_tensors="pt")
    output = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=True,
        temperature=0.9,
        top_k=50,
        pad_token_id=tokenizer.eos_token_id,
    )
    full_text = tokenizer.decode(output[0], skip_special_tokens=True)
    # Only return the NEW text (cut off the prompt part)
    return full_text[len(prompt):].strip()

# ==================================================
# GAME STATE
# ==================================================
player = {
    "name": "",
    "health": 100,
    "inventory": [],
}

# ==================================================
# SCENE FUNCTION
# ==================================================
def play_scene(scene_prompt, choices):
    """
    Generates a description for the current scene using GPT-2,
    then presents the player with a set of choices.
    """
    print("\n" + "=" * 60)
    description = generate_text(scene_prompt)
    print(description)
    print("=" * 60)

    for i, choice in enumerate(choices, start=1):
        print(f"{i}. {choice}")

    while True:
        try:
            pick = int(input("\nWhat do you do? (enter a number): "))
            if 1 <= pick <= len(choices):
                return pick
        except ValueError:
            pass
        print("Please enter a valid choice number.")

# ==================================================
# MAIN GAME LOOP
# ==================================================
def start_game():
    player["name"] = input("What is your hero's name? ")
    print(f"\nWelcome, {player['name']}! Your adventure begins...")

    # --- Scene 1 ---
    choice = play_scene(
        f"{player['name']} stood at the edge of a dark forest. The trees whispered",
        ["Enter the forest", "Turn back toward the village", "Call out to see if anyone answers"]
    )

    if choice == 1:
        next_prompt = f"{player['name']} stepped into the forest and immediately"
    elif choice == 2:
        next_prompt = f"{player['name']} turned back toward the village, but the path had changed and"
    else:
        next_prompt = f"{player['name']} called out into the dark forest, and a voice replied"

    # --- Scene 2 ---
    choice = play_scene(next_prompt, ["Keep moving forward", "Draw your weapon", "Hide and wait"])

    if choice == 1:
        outcome_prompt = f"{player['name']} kept moving forward, brave and determined, until"
    elif choice == 2:
        outcome_prompt = f"{player['name']} drew their weapon, ready for anything, when suddenly"
    else:
        outcome_prompt = f"{player['name']} hid quietly in the shadows, watching as"

    # --- Ending ---
    print("\n" + "=" * 60)
    print("THE FINAL CHAPTER")
    print("=" * 60)
    ending = generate_text(outcome_prompt, max_new_tokens=60)
    print(ending)
    print("\n--- THE END ---")

# ==================================================
# RUN THE GAME
# ==================================================
start_game()

# ==================================================
# CHALLENGE EXTENSIONS
# ==================================================
# 1. Add a health/combat system (lose health on certain choices).
# 2. Add an inventory system (pick up items, use them later).
# 3. Add more scenes and branches to make a longer story.
# 4. Try changing `temperature` (higher = wilder, lower = safer)
#    and see how it changes the vibe of the story.
```
